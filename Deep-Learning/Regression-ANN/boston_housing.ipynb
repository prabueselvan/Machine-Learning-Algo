## Data Preparation

```python
!pip install "numpy<2.0.0" "scipy<1.14.0" "scikit-learn<1.6.0"
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scalar =  StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
```

```python
X_train.mean().round(), X_train.std().round()
```

```python
X_test.mean().round(),X_test.std().round()
```

## Model Building

```python
from keras.models import Sequential
from keras.layers import Dense, Flatten 
```

```python
model=Sequential()
model.add(Dense (64, activation='relu', input_shape=(13,)))
model.add(Dense(32, activation='relu'))
model.add(Dense(16, activation='relu'))

# output layer
model.add(Dense(1))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='mse', metrics=["mae"])
```

```python
model.fit(X_train,y_train, epochs=16, validation_data=(X_test,y_test), verbose=2)
```

```python
y_pred = model.predict(X_test)
```
