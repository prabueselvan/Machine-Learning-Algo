## Model Building

```python
X = bread_cancer_data_df.drop(columns=["Target"], axis=1)
X.shape
```

```python
y= bread_cancer_data_df["Target"]
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=12, shuffle=True, stratify=bread_cancer_data_df.Target)
X_train, X_test, y_train, y_test 
```

```python
bread_cancer_data_df
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt = DecisionTreeClassifier(criterion='gini', max_depth=3, class_weight={0:2,1:1 })
dt_model = dt.fit(X_train, y_train)
dt_model
```

```python
plt.pie(x=y_train.value_counts(), autopct="%1.0f%%", labels=["Benign Cancer","Malignant Cancer"], explode=[0.01,0.01])
plt.show()
```

```python
plt.pie(x=y_test.value_counts(), autopct="%1.0f%%", labels=["Benign Cancer","Malignant Cancer"], explode=[0.01,0.01])
plt.show()
```

## 6. Model Testing || 7. Model Evaluation

```python
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print (classification_report(y_train, y_pred_train))
```

```python
print (confusion_matrix(y_train, y_pred_train))
```

```python
y_pred_test = dt_model.predict(X_test)
```

```python
print (classification_report(y_test, y_pred_test))
```

```python
print (confusion_matrix(y_test, y_pred_test))
```
