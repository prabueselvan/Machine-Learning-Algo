```python
X = california_full.drop(columns=["median_house_value"],axis=1)
X
```

```python
y = california_full["median_house_value"]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test =train_test_split(X, y , test_size=0.20,random_state=43)
```

```python
X_train, X_test, y_train, y_test
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lm=LinearRegression()
```

```python
lm_model = lm.fit(X_train, y_train)
```

```python
y_test_pred = lm_model.predict(X_test)
y_test_pred
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error, root_mean_squared_error,r2_score, roc_auc_score
```

```python
print(mean_absolute_error(y_test, y_test_pred))
print(mean_squared_error(y_test, y_test_pred))
print(root_mean_squared_error(y_test, y_test_pred))
print(r2_score(y_test, y_test_pred))
```

## HomoScadascity test

```python
residuals = y_test - y_test_pred
```

```python
print ("Error on Traning data --> ", lm_model.predict(X_train).mean())
print ("Error on Test data --> ", residuals.mean())
```

```python
from scipy import stats
```

```python
stats.ttest_1samp(residuals, 0)
```

```python
sns.scatterplot(x=y_test_pred, y= residuals, alpha=0.6)
plt.axhline(0, color='red',linestyle='--')
plt.xlabel("y_test_pred")
plt.ylabel("residuals")
plt.show()
```

```python
sns.histplot(residuals, kde=True)
```

## ROC Curve

```python
from sklearn.metrics import roc_curve, roc_auc_score

threshold = y_train.median()
y_test_bin = (y_test > threshold).astype(int)
fpr, tpr, thresholds = roc_curve(y_test_bin, y_test_pred)

auc = roc_auc_score(y_test_bin, y_test_pred)


plt.plot(fpr, tpr, label=f"Linear regression (AUC = {auc:3f})")
plt.plot([0,1], [0,1], "k--", label ="Random guess")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("ROC curve: expensive vs not expensive")
plt.legend(); plt.show()
```
