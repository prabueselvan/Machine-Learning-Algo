```python
# weather_data["Ozone"] = weather_data["Ozone"].fillna(weather_data["Ozone"].mean(), inplace=True)
# weather_data["Ozone"].dropna(axis=0, inplace=True)

del weather_data["Ozone"]
```

```python
weather_data.isna().sum()
```

```python
weather_data
```

```python
weather_data[weather_data["Solar.R"].isna()]
```

```python
weather_data.groupby("Weather")["Solar.R"].mean().round(2)
```

```python
weather_data["Solar.R"] =  weather_data.groupby("Weather")["Solar.R"].transform(lambda x:x.fillna(x.mean().round()))
```

```python
weather_data[ weather_data["Solar.R"].isna()] 
```

```python
weather_data.dropna(axis=0,inplace=True)
```

## Stage 2

## Data Transformation

* If the data is CONTINUOUS  --> Standard Scalar, MinMaxScalar, Robust Scalar
* If the data is DISCRETE --> Label Encoder , One HOT encoder

```python
weather_data_copy_01 = weather_data.copy()
weather_data_copy_01 
```

```python
from sklearn.preprocessing import LabelEncoder
le_encoder = LabelEncoder()
weather_data_copy_01["Weather"] = le_encoder.fit_transform(weather_data_copy_01["Weather"])
```

```python
weather_data_copy_01.head()
```

## One Hot Encoder

* pd.get_dummies
* sklearn - OneHotencoder

```python
weather_data_copy_02 = weather_data.copy()
weather_data_copy_02
```

```python
pd.get_dummies(data=weather_data_copy_02, columns=["Weather"], dtype=int)
```

```python
weather_data_copy_03 = weather_data.copy()
weather_data_copy_03
```

```python
from sklearn.preprocessing import OneHotEncoder

one_hot = OneHotEncoder(sparse_output=False).set_output(transform="pandas")
# one_hot = OneHotEncoder()
```

```python
encoded_weather = one_hot.fit_transform(weather_data_copy_03[["Weather"]])
encoded_weather

weather_data_copy_03 = weather_data_copy_03.drop(columns=["Weather"]).join(encoded_weather)
```
